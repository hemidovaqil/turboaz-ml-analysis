```python
import pandas as pd
import numpy as np

df = pd.read_csv("house_sale.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
print("Columns:")
print(df.columns.tolist())
```

```python
missing = df.isnull().sum()

print("Missing values:")
print(missing[missing > 0])
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().copy()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
print(df.dtypes)
```

```python
numeric_features = df.select_dtypes(include="number").columns.tolist()

categorical_features = df.select_dtypes(include="object").columns.tolist()

print("Numerical features:")
print(numeric_features)

print("Categorical features:")
print(categorical_features)
```

```python
for column in numeric_features:
      df[column] = df[column].fillna(df[column].median())

      for column in categorical_features:
          df[column] = df[column].fillna(df[column].mode()[0])

          print("Missing values remaining:", df.isnull().sum().sum())
```

```python
df[numeric_features].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.boxplot(data=df.select_dtypes(include="number"))

plt.title("Boxplots of Numerical Features")
plt.xticks(rotation=45)
plt.show()
```

## Control Point 2 — Data Cleaning and Preprocessing

The dataset was inspected for missing values, duplicate records, incorrect data types, and potential outliers. Duplicate rows were removed and missing values were handled according to the data type. Numerical and categorical features were identified for further preprocessing and machine learning.
