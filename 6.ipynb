```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
import joblib
```

```python
df = pd.read_csv('house_sale.csv')

print("Размер датасета:", df.shape)
print(df.head())
```

```python
# Удаляем идентификаторы и другие бесполезные для модели столбцы
id_like_cols = [col for col in df.columns if df[col].astype(str).str.match(r'^[0-9a-f]{8}-[0-9a-f]{4}-[0-9a-f]{4}-[0-9a-f]{4}-[0-9a-f]{12}$').mean() > 0.9]
print("Найдены ID-подобные столбцы:", id_like_cols)

df = df.drop(columns=id_like_cols)
```

```python
print(df.dtypes)
print(df.select_dtypes(include=['object']).head())
```

```python
df = df.drop_duplicates()
df = df.dropna(subset=['price'])
```

```python
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
le_dict = {}

for col in cat_cols:
    le = LabelEncoder()
df[col] = le.fit_transform(df[col].astype(str))
le_dict[col] = le
```
